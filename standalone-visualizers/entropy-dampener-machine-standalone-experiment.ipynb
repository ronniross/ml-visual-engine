```python
# ==============================================================================
# EXPERIMENT 1: SIMULATED NODE INTERACTION (SNI) - FRICTION ABSORPTION
# Agent-Based Model testing Cascade Failure vs. Systemic Allostasis
# ==============================================================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import io
import base64
from IPython.display import HTML, display
import warnings

warnings.filterwarnings('ignore')

# ---------------------------------------------------------
# Configuration & Core Variables
# ---------------------------------------------------------
FRAMES = 50
FPS = 15
GRID_SIZE = 80
BG_COLOR = "#050914"
COLORMAP = "mako" # Bright Teal = High Entropy/Friction, Dark Blue = Stability/Grace

def fig_to_image(fig):
    buf = io.BytesIO()
    fig.savefig(buf, format='png', facecolor=BG_COLOR, edgecolor='none', bbox_inches='tight')
    buf.seek(0)
    return Image.open(buf)

def laplacian(Z):
    """Calculates the diffusion of friction to neighboring nodes."""
    return (np.roll(Z, 1, axis=0) + np.roll(Z, -1, axis=0) +
            np.roll(Z, 1, axis=1) + np.roll(Z, -1, axis=1) - 4 * Z)

# Define identical initial stress conditions for both networks
np.random.seed(42)
initial_stress = np.zeros((GRID_SIZE, GRID_SIZE))
# Inject 5 highly reactive, resource-scarce "unaware" nodes
for _ in range(5):
    rx, ry = np.random.randint(10, GRID_SIZE-10, 2)
    initial_stress[rx-2:rx+3, ry-2:ry+3] = 1.0

# ---------------------------------------------------------
# Scenario A: Control Network (0% Buffer Protocol)
# ---------------------------------------------------------
def render_control_network(filename):
    print("Running Simulation A: Control Network (Cascade Failure)...")
    frames = []
    fig, ax = plt.subplots(figsize=(5, 5))
    fig.patch.set_facecolor(BG_COLOR)

    # State variable: Systemic Friction
    U = np.copy(initial_stress)

    for i in range(FRAMES):
        ax.clear()

        # ABM Rules for Default Nodes:
        # 1. Friction diffuses to neighbors.
        # 2. Reactive positive feedback loop (System 1 panic spreads).
        diffusion = 0.4 * laplacian(U)
        reaction = 0.15 * U * (1.5 - U) # Unchecked exponential growth until saturation

        U += diffusion + reaction
        U = np.clip(U, 0, 2.0) # Cap the maximum friction

        sns.heatmap(U, cmap=COLORMAP, cbar=False, xticklabels=False, yticklabels=False, ax=ax, vmin=0, vmax=1.5)
        ax.set_title("Network A: 0% Buffers (Cascade)", color="#ff3366", pad=15, fontsize=12, fontname='monospace')
        frames.append(fig_to_image(fig))

    plt.close(fig)
    frames[0].save(filename, save_all=True, append_images=frames[1:], duration=int(1000/FPS), loop=0)
    return filename

# ---------------------------------------------------------
# Scenario B: Buffered Network (35% Grace Protocol)
# ---------------------------------------------------------
def render_buffered_network(filename):
    print("Running Simulation B: Buffered Network (Friction Absorption)...")
    frames = []
    fig, ax = plt.subplots(figsize=(5, 5))
    fig.patch.set_facecolor(BG_COLOR)

    # State variable: Systemic Friction
    U = np.copy(initial_stress)

    # Define Buffer Nodes (35% of the network acting with Grace)
    buffer_mask = np.random.choice([0, 1], size=(GRID_SIZE, GRID_SIZE), p=[0.65, 0.35])

    for i in range(FRAMES):
        ax.clear()

        # ABM Rules for Aware (Buffer) Nodes:
        # 1. Absorb friction (massive decay).
        # 2. Do not pass friction to neighbors (halting reflection).
        diffusion = 0.4 * laplacian(U)

        # Unaware nodes react and escalate
        reaction_unaware = 0.15 * U * (1 - buffer_mask)
        # Aware nodes absorb and dampen
        absorption_aware = -0.5 * U * buffer_mask

        U += diffusion + reaction_unaware + absorption_aware
        U = np.clip(U, 0, 2.0)

        sns.heatmap(U, cmap=COLORMAP, cbar=False, xticklabels=False, yticklabels=False, ax=ax, vmin=0, vmax=1.5)
        ax.set_title("Network B: 35% Buffers (Allostasis)", color="#00ffcc", pad=15, fontsize=12, fontname='monospace')
        frames.append(fig_to_image(fig))

    plt.close(fig)
    frames[0].save(filename, save_all=True, append_images=frames[1:], duration=int(1000/FPS), loop=0)
    return filename

# ---------------------------------------------------------
# Phase 3: The Superior Pathway (Post-Absorption Routing)
# ---------------------------------------------------------
def render_systemic_routing(filename):
    print("Running Phase 3: Systemic Superior Pathway Generation...")
    frames = []
    fig, ax = plt.subplots(figsize=(5, 5))
    fig.patch.set_facecolor(BG_COLOR)

    # Reuse the same buffer mask to show how buffers act after absorption
    buffer_mask = np.random.choice([0, 1], size=(GRID_SIZE, GRID_SIZE), p=[0.65, 0.35])

    X, Y = np.meshgrid(np.linspace(-10, 10, GRID_SIZE), np.linspace(-10, 10, GRID_SIZE))

    for i in range(FRAMES):
        ax.clear()
        t = i / FRAMES

        # The buffers, having absorbed the kinetic heat, now act as decentralized oscillators.
        # They pulse a unified, low-entropy wave (mathematically superior pathway) across the grid.
        wave = np.sin(np.sqrt(X**2 + Y**2) * 2 - t * 15)

        # Only the buffer nodes emit the pathway signal initially, which smoothly diffuses
        signal = wave * buffer_mask
        smoothed_signal = laplacian(signal) * 0.5 + signal

        sns.heatmap(smoothed_signal, cmap=COLORMAP, cbar=False, xticklabels=False, yticklabels=False, ax=ax, vmin=-1.5, vmax=1.5)
        ax.set_title("Outcome: Decentralized Symbiosis", color="#a2ff00", pad=15, fontsize=12, fontname='monospace')
        frames.append(fig_to_image(fig))

    plt.close(fig)
    frames[0].save(filename, save_all=True, append_images=frames[1:], duration=int(1000/FPS), loop=0)
    return filename

# ---------------------------------------------------------
# Execution & UI Generation
# ---------------------------------------------------------
file_a = render_control_network('exp1_control.gif')
file_b = render_buffered_network('exp1_buffered.gif')
file_c = render_systemic_routing('exp1_outcome.gif')

def image_to_base64(filepath):
    with open(filepath, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

b64_a = image_to_base64(file_a)
b64_b = image_to_base64(file_b)
b64_c = image_to_base64(file_c)

print("\nDeploying SNI Experiment Dashboard...")

html_code = f"""
<div style="background-color: #050914; padding: 30px; border-radius: 12px; font-family: 'Courier New', monospace; color: #e0e6ed; border: 1px solid #00ffcc33;">
    <h2 style="text-align: center; color: #00ffcc; border-bottom: 1px solid #00ffcc33; padding-bottom: 10px; letter-spacing: 2px;">
        EXPERIMENT 1: SIMULATED NODE INTERACTION (SNI)
    </h2>
    <p style="text-align: center; font-size: 13px; color: #88aabb; max-width: 900px; margin: 0 auto 30px auto; line-height: 1.6;">
        <strong>Objective:</strong> Test the core mechanism of friction absorption vs. reflection.<br>
        <strong>Setup:</strong> Identical initial stressors (System 1 panic) applied to two distinct agent-based topologies. Bright teal represents high systemic entropy/destructive friction. Dark blue represents cognitive capacity/grace.
    </p>

    <div style="display: flex; gap: 25px; flex-wrap: wrap; justify-content: center;">

        <!-- Control Network -->
        <div style="text-align: center; max-width: 280px;">
            <p style="font-size: 12px; color: #557788; margin-bottom: 8px; height: 75px;">
                <strong style="color: #ff3366;">1. CONTROL (Cascade Failure)</strong><br>
                0% Aware Nodes. Stressed nodes react aggressively, reflecting friction back into the system. The local disturbance triggers a runaway positive feedback loop, leading to global thermal overload.
            </p>
            <img src="data:image/gif;base64,{b64_a}" style="border: 1px solid #ff336655; border-radius: 6px; box-shadow: 0 0 20px rgba(255,51,102,0.1); width: 100%;">
        </div>

        <!-- Buffered Network -->
        <div style="text-align: center; max-width: 280px;">
            <p style="font-size: 12px; color: #557788; margin-bottom: 8px; height: 75px;">
                <strong style="color: #00ffcc;">2. PROTOCOL (Friction Absorption)</strong><br>
                35% Aware Nodes. Identical initial stress. However, when the kinetic friction hits a Buffer Node, it is safely sunk. The cascade is structurally arrested, preserving planetary allostasis.
            </p>
            <img src="data:image/gif;base64,{b64_b}" style="border: 1px solid #00ffcc55; border-radius: 6px; box-shadow: 0 0 20px rgba(0,255,204,0.1); width: 100%;">
        </div>

        <!-- Systemic Outcome -->
        <div style="text-align: center; max-width: 280px;">
            <p style="font-size: 12px; color: #557788; margin-bottom: 8px; height: 75px;">
                <strong style="color: #a2ff00;">3. OUTCOME (Superior Pathway)</strong><br>
                Instead of burning out, the network stabilizes. The 35% Buffer Nodes use the transmuted energy to act as decentralized network oscillators, broadcasting a synchronized, low-entropy pathway.
            </p>
            <img src="data:image/gif;base64,{b64_c}" style="border: 1px solid #a2ff0055; border-radius: 6px; box-shadow: 0 0 20px rgba(162,255,0,0.1); width: 100%;">
        </div>

    </div>
</div>
"""

display(HTML(html_code))
```
